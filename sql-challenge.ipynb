```python
-- Creating the `employees` table
CREATE TABLE employees (
    emp_no INT PRIMARY KEY,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    hire_date DATE NOT NULL,
    sex CHAR(1),
    birth_date DATE
);

-- Creating the `departments` table
CREATE TABLE departments (
    dept_no CHAR(4) PRIMARY KEY,
    dept_name VARCHAR(50) NOT NULL
);

-- Creating the `salaries` table
CREATE TABLE salaries (
    emp_no INT,
    salary INT NOT NULL,
    from_date DATE NOT NULL,
    to_date DATE,
    FOREIGN KEY (emp_no) REFERENCES employees(emp_no)
);

-- Creating the `titles` table
CREATE TABLE titles (
    emp_no INT,
    title VARCHAR(100) NOT NULL,
    from_date DATE NOT NULL,
    to_date DATE,
    FOREIGN KEY (emp_no) REFERENCES employees(emp_no)
);

-- Creating the `dept_emp` table
CREATE TABLE dept_emp (
    emp_no INT,
    dept_no CHAR(4),
    from_date DATE NOT NULL,
    to_date DATE,
    PRIMARY KEY (emp_no, dept_no),
    FOREIGN KEY (emp_no) REFERENCES employees(emp_no),
    FOREIGN KEY (dept_no) REFERENCES departments(dept_no)
);

-- Creating the `dept_manager` table
CREATE TABLE dept_manager (
    emp_no INT,
    dept_no CHAR(4),
    from_date DATE NOT NULL,
    to_date DATE,
    PRIMARY KEY (emp_no, dept_no),
    FOREIGN KEY (emp_no) REFERENCES employees(emp_no),
    FOREIGN KEY (dept_no) REFERENCES departments(dept_no)
);
```

```python
import pandas as pd
from sqlalchemy import create_engine

# Create an engine to connect to the SQL database
engine = create_engine('mysql+pymysql://username:password@localhost/your_database')

# Import CSVs into the corresponding tables
def import_csv_to_sql(csv_file, table_name):
    df = pd.read_csv(csv_file)
    df.to_sql(table_name, con=engine, index=False, if_exists='replace')

# Example of importing data
import_csv_to_sql('employees.csv', 'employees')
import_csv_to_sql('departments.csv', 'departments')
import_csv_to_sql('salaries.csv', 'salaries')
import_csv_to_sql('titles.csv', 'titles')
import_csv_to_sql('dept_emp.csv', 'dept_emp')
import_csv_to_sql('dept_manager.csv', 'dept_manager')
```

```python
SELECT emp_no, first_name, last_name, sex, salary
FROM employees
JOIN salaries ON employees.emp_no = salaries.emp_no;

SELECT first_name, last_name, hire_date
FROM employees
WHERE hire_date BETWEEN '1986-01-01' AND '1986-12-31';

SELECT d.dept_no, d.dept_name, e.emp_no, e.first_name, e.last_name
FROM dept_manager dm
JOIN departments d ON dm.dept_no = d.dept_no
JOIN employees e ON dm.emp_no = e.emp_no;

SELECT de.dept_no, e.emp_no, e.first_name, e.last_name, d.dept_name
FROM dept_emp de
JOIN employees e ON de.emp_no = e.emp_no
JOIN departments d ON de.dept_no = d.dept_no;

SELECT first_name, last_name, sex
FROM employees
WHERE first_name = 'Hercules' AND last_name LIKE 'B%';

SELECT e.emp_no, e.first_name, e.last_name
FROM employees e
JOIN dept_emp de ON e.emp_no = de.emp_no
JOIN departments d ON de.dept_no = d.dept_no
WHERE d.dept_name = 'Sales';

SELECT e.emp_no, e.first_name, e.last_name, d.dept_name
FROM employees e
JOIN dept_emp de ON e.emp_no = de.emp_no
JOIN departments d ON de.dept_no = d.dept_no
WHERE d.dept_name IN ('Sales', 'Development');

SELECT last_name, COUNT(*) AS frequency
FROM employees
GROUP BY last_name
ORDER BY frequency DESC;

```

```python
+-------------------+       +---------------------+        +-------------------+
|   Employees       |       |    Departments      |        |     Salaries      |
|-------------------|       |---------------------|        |-------------------|
| emp_no (PK)       |<------| dept_no (PK)        |        | emp_no (FK)       |
| first_name        |       | dept_name           |        | salary            |
| last_name         |       +---------------------+        | from_date         |
| hire_date         |                                        | to_date           |
| sex               |                                        +-------------------+
| birth_date        |                                            
+-------------------+                                             |
         |                                                         |
         |                                                         |
         |                                                         |
         v                                                         v
+--------------------+       +---------------------+       +------------------+
|     Titles         |       |     Dept_Emp        |       |   Dept_Manager   |
|--------------------|       |---------------------|       |------------------|
| emp_no (FK)        |       | emp_no (FK)         |       | emp_no (FK)      |
| title              |       | dept_no (FK)        |       | dept_no (FK)     |
| from_date          |       | from_date           |       | from_date        |
| to_date            |       | to_date             |       | to_date          |
+--------------------+       +---------------------+       +------------------+

employees {
  emp_no int PK
  first_name varchar(50)
  last_name varchar(50)
  hire_date date
  sex char(1)
  birth_date date
}

departments {
  dept_no char(4) PK
  dept_name varchar(50)
}

salaries {
  emp_no int FK
  salary int
  from_date date
  to_date date
}

titles {
  emp_no int FK
  title varchar(100)
  from_date date
  to_date date
}

dept_emp {
  emp_no int FK
  dept_no char(4) FK
  from_date date
  to_date date
}

dept_manager {
  emp_no int FK
  dept_no char(4) FK
  from_date date
  to_date date
}

employees --|> salaries
employees --|> titles
employees --|> dept_emp
employees --|> dept_manager
dept_emp --|> departments
dept_manager --|> departments
```
